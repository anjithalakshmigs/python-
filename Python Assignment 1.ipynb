```python
import numpy as np
import pandas as pd

# ==========================================
# PART 1: NumPy Array Operations
# ==========================================

# 1. Create a 1D NumPy Array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("1. temperatures_w1:", temperatures_w1)

# 2. Inspection and Properties
print("\n--- 2. Inspection and Properties ---")
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)

# 3. Array Operations
print("\n--- 3. Array Operations ---")
fahrenheit = (temperatures_w1 * 9 / 5) + 32
print("Fahrenheit temperatures:", fahrenheit)
print("Maximum temperature:", np.max(temperatures_w1))
print("Minimum temperature:", np.min(temperatures_w1))
print("Mean temperature:", np.mean(temperatures_w1))

# 4. Array Slicing and Indexing
print("\n--- 4. Array Slicing and Indexing ---")
first_three_days = temperatures_w1[:3]
weekend = temperatures_w1[-2:]
middle_three_days = temperatures_w1[2:5]
print("First three days:", first_three_days)
print("Weekend (last two days):", weekend)
print("Middle three days:", middle_three_days)

# 5. Create a 2D Array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1],
])

# 6. Inspect and Slice the 2D Array
print("\n--- 6. Inspect and Slice 2D Array ---")
print("2D Array Shape:", temperatures.shape)
print("2D Array Data Type:", temperatures.dtype)
print("2D Array Total Number of Elements:", temperatures.size)

week1_temps = temperatures[0, :]
week2_temps = temperatures[1, :]
weekends_both_weeks = temperatures[:, -2:]

print("Week 1 temperatures:", week1_temps)
print("Week 2 temperatures:", week2_temps)
print("Weekends for both weeks:\n", weekends_both_weeks)


# ==========================================
# PART 2: Pandas Series
# ==========================================

# 1. Creating Pandas Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'],
    name='marks',
)

# 2. Indexing and Slicing
print("\n--- Pandas Series: Indexing and Slicing ---")
mark_rank1 = marks.iloc[0]  # or marks[0]
top_3_ranks = marks.loc[['Rank1', 'Rank2', 'Rank3']]
mark_rank3 = marks.iloc[2]
marks_above_90 = marks[marks > 90]

print("Mark of 1st rank student:", mark_rank1)
print("Marks of top 3 ranks:\n", top_3_ranks)
print("Mark of 3rd rank student:", mark_rank3)
print("Ranks with marks > 90:\n", marks_above_90)

# 3. Manipulating Series
print("\n--- Pandas Series: Manipulating Series ---")
marks['Rank1'] = 100
print("Modified mark for 1st rank:\n", marks)

marks = marks.drop('Rank5')
print("After removing last rank:\n", marks)

cgpa = marks / 10
print("CGPA:\n", cgpa)


# ==========================================
# PART 3: Pandas DataFrame
# ==========================================

# 1. Creating Pandas DataFrame
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics',
        'Clothing',
        'Electronics',
        'Furniture',
        'Clothing',
        'Electronics',
        'Furniture',
        'Clothing',
        'Furniture',
        'Electronics',
    ],
    'Region': [
        'North',
        'South',
        'North',
        'East',
        'West',
        'North',
        'East',
        'West',
        'South',
        'North',
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400],
}

transactions = pd.DataFrame(data)

# 2. Data Exploration
print("\n--- Pandas DataFrame: Data Exploration ---")
print("First 5 rows (head):\n", transactions.head())
print("Last 5 rows (tail):\n", transactions.tail())
print("Shape:", transactions.shape)
print("Column names:", transactions.columns.tolist())
print("\nData info:")
transactions.info()

print("\nOnly 'ProductCategory' and 'Amount' columns:\n")
print(transactions[['ProductCategory', 'Amount']])

print("\nLast 3 columns using iloc:\n")
print(transactions.iloc[:, -3:])

filtered_rows = transactions[
    (transactions['Region'] == 'North') & (transactions['Amount'] > 200)
]
print("\nFiltered Rows (Region == 'North' and Amount > 200):\n", filtered_rows)

print(
    "\nValue counts for ProductCategory:\n",
    transactions['ProductCategory'].value_counts(),
)
print("Unique values in Region:", transactions['Region'].unique())
print(
    "\nMean amount by Region:\n", transactions.groupby('Region')['Amount'].mean()
)

# 3. Manipulating the DataFrame
print("\n--- Pandas DataFrame: Manipulating DataFrame ---")

# Modify Amount for TransactionID 102
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

# Add Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

# Remove row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

# Delete 'Discount' column
transactions = transactions.drop(columns=['Discount'])

print("\nFinal DataFrame:\n", transactions)
```
